```python
# main.py
from pyspark.sql import SparkSession

# Import all functions from the modular files
from data_loader import load_hr_data
from preprocessing import run_preprocessing_pipeline
from eda_analysis import analyze_attrition_distribution, analyze_salary_vs_performance, analyze_department_insights
from feature_engineering import create_indicators
from models_classification import build_attrition_model, build_performance_model
from models_clustering import segment_employees

if __name__ == "__main__":
    print("--- Starting AI-Powered HR Analytics Pipeline --- \n")

    # 1. Initialize SparkSession
    # This is essential in VS Code to use PySpark locally
    spark = SparkSession.builder \
        .appName("HR_Analytics_Project") \
        .getOrCreate()

    # 2. Define the path to the dataset
    # Since you are on VS Code, ensure the CSV is in the same directory
    file_path = "HR-Employee-Attrition.csv" 
    
    # 3. Load the data
    raw_df = load_hr_data(spark, file_path)

    # 4. Clean and preprocess the data
    clean_df = run_preprocessing_pipeline(raw_df)
    
    print("Preview of Cleaned Data:")
    # Use show() instead of display() in VS Code
    clean_df.show(5) 

    # 5. Exploratory Data Analysis (EDA)
    attrition_df = analyze_attrition_distribution(clean_df)
    attrition_df.show()

    salary_perf_df = analyze_salary_vs_performance(clean_df)
    salary_perf_df.show()

    dept_insights_df = analyze_department_insights(clean_df)
    dept_insights_df.show()

    # 6. Feature Engineering
    feature_df = create_indicators(clean_df)
    
    print("Preview of Engineered Features:")
    feature_df.select("Department", "Engagement_Indicator", "Loyalty_Indicator", "Income_Experience_Ratio").show(5)

    # 7. Train AI Models
    # A. Classification Model (Attrition Prediction)
    attrition_model, attrition_predictions = build_attrition_model(feature_df)

    # B. Regression Model (Performance Prediction)
    performance_model, performance_predictions = build_performance_model(feature_df)

    # C. Clustering Model (Employee Segmentation)
    cluster_predictions = segment_employees(feature_df)

    print("Preview of Employee Clusters:")
    cluster_predictions.select("Age", "MonthlyIncome", "Engagement_Indicator", "prediction").show(10)

    print("\n --- Pipeline Execution Completed Successfully! ---")
    
    # Stop the Spark session to free up resources
    spark.stop()
```
